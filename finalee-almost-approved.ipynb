````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from langchain.llms.base import LLM
from typing import Any

MODEL_NAME = "mistralai/Mistral-Nemo-Instruct-2407"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)
````

````python
def generate_text(prompt, max_length=4000, num_return_sequences=1):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(
        **inputs,
        max_length=max_length,
        num_return_sequences=num_return_sequences,
        do_sample=True,
        top_k=50,
        top_p=0.95,
        temperature=0.7,
    )
    return [tokenizer.decode(output, skip_special_tokens=True) for output in outputs][0]
````

````python
# LangChain wrapper
class CustomHFLLM(LLM):
    def _call(self, prompt: str, stop: Any = None) -> str:
        return generate_text(prompt, max_length=2600)
    
    @property
    def _llm_type(self):
        return "custom_huggingface"

llm = CustomHFLLM()
````

````python
import re
import json

def extract_json_block(text: str) -> str:
    """
    Extract the last JSON block from a text string.
    Looks for ```json ... ``` blocks and returns the content.
    """
    pattern = r"```json\s*(.*?)\s*```"
    matches = re.findall(pattern, text, re.DOTALL)
    if not matches:
        raise ValueError("No JSON block found in text.")
    return matches[-1]  # return last JSON block

def parse_json_from_text(text: str) -> dict:
    """
    Extract JSON block and convert to Python dict.
    """
    json_text = extract_json_block(text)
    try:
        return json.loads(json_text)
    except json.JSONDecodeError as e:
        # fallback: try to fix common issues
        start = json_text.find("{")
        end = json_text.rfind("}") + 1
        return json.loads(json_text[start:end])
````

````python
import pandas as pd

def load_excel_data(path: str):
    """Load mentors and interns from Excel and convert to lists of dicts."""

    df_mentors = pd.read_excel(path, sheet_name="Mentors")
    df_interns = pd.read_excel(path, sheet_name="Interns")

    # Convert "skills" from "x, y, z" → ["x","y","z"]
    df_mentors["skills"] = df_mentors["skills"].apply(lambda x: [s.strip() for s in x.split(",")])
    df_interns["skills"] = df_interns["skills"].apply(lambda x: [s.strip() for s in x.split(",")])

    mentors_list = df_mentors.to_dict(orient="records")
    interns_list = df_interns.to_dict(orient="records")

    return mentors_list, interns_list
````

````python
# Project Plan Parser
from langchain.output_parsers import StructuredOutputParser, ResponseSchema
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
````

````python
import json
````

````python
def generate_project_plan(project_title: str, mentors_list):
    project_plan_schemas = [
    ResponseSchema(name="title", description="Provide the actual name of the internship track/program."),
    ResponseSchema(name="duration", description="Provide actual start and end dates in the format YYYY-MM-DD."),
    ResponseSchema(name="objectives", description="Provide a list of 5 SMART objectives, NOT the word 'string'."),
    ResponseSchema(name="milestones", description="Provide 3–6 milestones, each as an object with name, due_date, and deliverables."),
    ResponseSchema(name="tasks", description="Provide a list of tasks with actual data: title, description, due_date, dependencies, assigned_mentor, required_resources."),
    ResponseSchema(name="evaluation", description="Provide real evaluation metrics, checkpoints, and feedback approaches."),
    ResponseSchema(name="risks", description="Provide real risks and their mitigations.")
]
    project_plan_parser = StructuredOutputParser.from_response_schemas(project_plan_schemas)
    project_plan_instructions = project_plan_parser.get_format_instructions()

    plan_project_prompt = PromptTemplate(
    input_variables=["project_title", "mentors_db", "project_plan_instructions"],
    template="""
    Generate a structured internship program plan in STRICT JSON format for the internship track/project: "{project_title}".
    Use the following mentors to assign tasks wherever possible: {mentors_db}.

    {project_plan_instructions}
    """
    )
    plan_project = LLMChain(llm=llm, prompt=plan_project_prompt)
    mentors_str = str(mentors_list)
    project_plan_text = plan_project.run({
        "project_title": project_title,
        "mentors_db": mentors_str,
        "project_plan_instructions":project_plan_instructions
    })
    project_plan_json = extract_json_block(project_plan_text)
    return project_plan_json
````

````python
def generate_task_assignments(project_plan_json):
    
    # ---------------------
    # STEP 1: Extract tasks
    # ---------------------
    task_list_schema = [
       ResponseSchema(name="tasks",
       description="A JSON array of beginner-friendly internship tasks. "
                "Each task MUST be an object containing: title, description, difficulty. "
                "Do NOT output the word 'string'."
)
    ]
    task_list_parser = StructuredOutputParser.from_response_schemas(task_list_schema)
    task_list_instructions = task_list_parser.get_format_instructions()

    task_list_prompt = PromptTemplate(
        input_variables=["project_plan", "format_instructions"],
        template="""
        your task is GENERATE List of beginner-friendly internship tasks based on the project plan below.  
        Return ONLY JSON.

        Project Plan:
        {project_plan}

        {format_instructions}
        """
        )

    task_list_chain = LLMChain(llm=llm, prompt=task_list_prompt)
    task_list_text = task_list_chain.run({
        "project_plan": str(project_plan_json),
        "format_instructions": task_list_instructions
    })

    tasks_json=extract_json_block(task_list_text)
    return tasks_json
````

````python
def run_mentor_matching(interns_list, mentors_list):
    # Mentor Matching Parser
    mentor_schemas = [
    ResponseSchema(name="mentor_name", description="Name of the best matching mentor"),
    ResponseSchema(name="interns_ids_list", description="list of interns that best matching mentor"),
    ResponseSchema(name="justification", description="Concise reasoning referencing skill overlaps")
    ]
    mentor_parser = StructuredOutputParser.from_response_schemas(mentor_schemas)
    mentor_instructions = mentor_parser.get_format_instructions()
    #Generate Prompt
    mentor_prompt_template = PromptTemplate(
    input_variables=["intern_skills", "mentors_db", "mentor_instructions"],
    template="""
    Pick exactly ONE mentor from the provided list whose skills best match each intern 
    (or GROUP interns under mentors if multiple interns match ).
    
    Return STRICT JSON .

    Intern skills: {intern_skills}
    Mentor profiles: {mentors_db}

    {mentor_instructions}
    """
    )
    mentor_chain = LLMChain(llm=llm, prompt=mentor_prompt_template)
    interns_str = str(interns_list)
    mentors_str = str(mentors_list)
    mentor_text = mentor_chain.run({
        "intern_skills": interns_str,
        "mentors_db": mentors_str,
        "mentor_instructions":mentor_instructions
    })
    mentor_json = extract_json_block(mentor_text)
    return mentor_json
````

````python
excel_path="/kaggle/input/finale/interns_mentors_dataset.xlsx"
project_title="Web development Internship"
mentors_db, interns_db = load_excel_data(excel_path)
project_plan = generate_project_plan(project_title, mentors_db)
print("\n--- Project Plan ---")
print(project_plan)
task_assignments = generate_task_assignments(project_plan)
print("\n--- Task Assignments ---")
print(task_assignments)
mentor_match = run_mentor_matching(interns_db, mentors_db)
print("\n--- Mentor Match ---")
print(mentor_match)
````

````python
from fastapi import FastAPI, UploadFile, File, HTTPException, Form
from fastapi.middleware.cors import CORSMiddleware
import pandas as pd
import tempfile
import shutil

# Your other functions must already exist:
# generate_project_plan()
# generate_task_assignments()
# run_mentor_matching()

app = FastAPI()

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

def load_excel_data_from_upload(file_path: str):
    """Load mentors and interns from Excel and convert to lists of dicts."""
    df_mentors = pd.read_excel(file_path, sheet_name="Mentors")
    df_interns = pd.read_excel(file_path, sheet_name="Interns")

    df_mentors["skills"] = df_mentors["skills"].apply(lambda x: [s.strip() for s in x.split(",")])
    df_interns["skills"] = df_interns["skills"].apply(lambda x: [s.strip() for s in x.split(",")])

    mentors_list = df_mentors.to_dict(orient="records")
    interns_list = df_interns.to_dict(orient="records")

    return mentors_list, interns_list


@app.post("/process-project")
async def process_project(
    project_title: str = Form(...),
    file: UploadFile = File(...)
):
    """
    Upload Excel → extract mentors+interns → generate project plan, tasks & mentor matches.
    """

    try:
        # Save uploaded Excel temporarily
        with tempfile.NamedTemporaryFile(delete=False, suffix=".xlsx") as tmp:
            shutil.copyfileobj(file.file, tmp)
            temp_path = tmp.name

        # Load mentors & interns directly from uploaded Excel
        mentors_db, interns_db = load_excel_data_from_upload(temp_path)

        # Generate results
        project_plan = generate_project_plan(project_title, mentors_db)
        task_assignments = generate_task_assignments(project_plan)
        mentor_match = run_mentor_matching(interns_db, mentors_db)

        return {
            "project_plan": project_plan,
            "task_assignments": task_assignments,
            "mentor_match": mentor_match
        }

    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))
````
